**2. Арифметичні операції**

Створіть дві змінні: `var1` зі значенням 15, `var2` зі значенням 4.

Обчисліть та виведіть (кожен пункт оцінюється в 0.5 бала):
- Суму
- Різницю
- Добуток
- Ділення
- Цілочисельне ділення
- Залишок від ділення
- Піднесення до степеня


```python
var1 = 15
var2 = 4

sum = var1 + var2
difference = var1 - var2
multiplication = var1 * var2
division = var1 / var2
integer_division = var1 // var2
remainder = var1 % var2
exponentiation = var1 ** var2

print(f"сумма - {sum};",
      f"різниця - {difference};",
      f"добуток - {multiplication};",
      f"ділення - {division};",
      f"цілочисельне ділення - {integer_division};",
      f"залишок від ділення - {remainder};",
      f"піднесення до степеня - {exponentiation}")
```
